# TickBytes Quick Start Example

This notebook demonstrates how to load, analyze, and query the **TickBytes** daily tick-by-tick (Level 1 & Level 2) datasets using Python, Pandas, and Polars.

```python
import pandas as pd
import polars as pl
import numpy as np


from pathlib import Path

# Set file paths
spot_file = Path("../../samples/spot/NIFTY_spot.csv")
fut_file = Path("../../samples/futures/NIFTY_FUT.csv")
opt_file = Path("../../samples/options/NIFTY_OPT.csv")
```

## 1. Load Spot Index Ticks with Pandas

```python
df_spot = pd.read_csv(spot_file)
df_spot['datetime'] = pd.to_datetime(df_spot['datetime'])
df_spot.head()
```

## 2. Load Futures Ticks & Analyze Level 2 Market Depth

```python
df_fut = pd.read_csv(fut_file)
df_fut['datetime'] = pd.to_datetime(df_fut['datetime'])

# Calculate Bid-Ask Spread at each tick
df_fut['spread'] = df_fut['ask_px'] - df_fut['bid_px']
print(f"Average FUT Bid-Ask Spread: {df_fut['spread'].mean():.2f} points")
```

## 3. Load Options Ticks & Analyze Implied Volatility and Greeks

```python
df_opt = pd.read_csv(opt_file)
df_opt['datetime'] = pd.to_datetime(df_opt['datetime'])
print(f"Average Implied Volatility: {df_opt['iv'].mean():.4%}")
print(f"Average Option Delta: {df_opt['delta'].mean():.4f}")
```
